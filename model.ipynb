```python
model_df = df.groupby(["player", "period"]).last().reset_index()
model_df = model_df.sort_values(by=["player", "season", "period"])
model_df["previous_period_zscore"] = model_df["bpm_zscore"].shift(1)
model_df["next_player"] = model_df["player"].shift(-1)
model_df.loc[
    model_df["player"] != model_df["next_player"], "previous_period_zscore"
] = float("nan")
model_df["difference"] = -model_df["previous_period_zscore"] + model_df["bpm_zscore"]
model_df.drop(columns="next_player", inplace=True)
model_df["comeback_from_injury"] = model_df["comeback_from_injury"].fillna("None")
model_df = model_df[
    [
        "previous_period_zscore",
        "bpm_zscore",
        "difference",
        "comeback_from_injury",
        "Age",
    ]
]
model_df = model_df.dropna().reset_index(drop=True)
model_df
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

results = []

# Assume 'difference', 'previous_period_zscore', 'injury_type' are in your DataFrame
for injury in model_df["comeback_from_injury"].dropna().unique():
    subset = model_df[model_df["comeback_from_injury"] == injury]

    # X = previous z-score, y = difference
    X = subset[["previous_period_zscore", "Age"]]
    y = subset["bpm_zscore"]

    # Drop rows where X or y is NaN or inf
    valid = (
        X.notnull().all(axis=1)
        & y.notnull()
        & np.isfinite(X).all(axis=1)
        & np.isfinite(y)
    )
    X = X[valid]
    y = y[valid]

    if len(X) == 0:
        continue  # Skip if no valid data

    # Fit sklearn LinearRegression (automatically handles intercept)
    model = LinearRegression()
    model.fit(X, y)

    r2 = model.score(X, y)
    y_pred = model.predict(X)
    rmse = np.sqrt(mean_squared_error(y, y_pred))
    coef = model.coef_[0]
    intercept = model.intercept_

    results.append(
        {
            "comeback_from_injury": injury,
            "r2": r2,
            "rmse": rmse,
            "coef": coef,
            "intercept": intercept,
            "n_samples": len(subset),
        }
    )

# Convert to DataFrame for viewing
results_df = pd.DataFrame(results).sort_values("r2", ascending=False)
results_df
```

```python
import statsmodels.api as sm
import pandas as pd

# Define X and y
X = model_df_dummies[
    ["previous_period_zscore", "Age"]
    + [
        col
        for col in model_df_dummies.columns
        if col not in ["previous_period_zscore", "Age", "bpm_zscore", "difference"]
    ]
]
X = sm.add_constant(X)
y = model_df_dummies["bpm_zscore"]

# Fit the model using statsmodels
model = sm.OLS(y, X).fit()

# Get p-values and coefficients
summary_df = (
    model.summary2().tables[1].reset_index()
)  # easier to manipulate than summary()
summary_df = summary_df.rename(
    columns={
        "index": "feature",
        "Coef.": "coefficient",
        "P>|t|": "p_value",
        "Std.Err.": "std_error",
        "t": "t_stat",
    }
)

# Sort by coefficient like your sklearn version
summary_df = summary_df.sort_values(by="coefficient")
summary_df["p_value"] = summary_df["p_value"].apply(
    lambda x: f"{float(x):.4f}" if pd.notnull(x) else "NaN"
)

(summary_df[["feature", "coefficient", "std_error", "t_stat", "p_value"]])
```

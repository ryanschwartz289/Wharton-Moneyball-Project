```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
import time

tables = []
for i in range(2010, 2025):
    time.sleep(0.1)
    url = f"https://www.basketball-reference.com/leagues/NBA_{i}_play-by-play.html"
    response = requests.get(url)
    soup = BeautifulSoup(response.content, "html.parser")
    table = soup.find("table", {"id": "pbp_stats"})
    if table:
        table_df = pd.read_html(str(table))[0]
        table_df.columns = table_df.columns.droplevel(
            0
        )  # Drop the first level of the MultiIndex
        tables.append(table_df)
```

```python
for i, table_df in enumerate(tables):
    season = 2010 + i
    table_df["season"] = season
    table_df.drop(columns=["seasons"])
```

```python
df = pd.concat(tables, ignore_index=True)
df = df[["Player", "Age", "season"]]
df
```

```python
df = df.drop_duplicates(subset=["Player", "season"], keep="last")
df.to_csv("./data/player_ages.csv", index=False)
```

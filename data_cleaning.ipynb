## Creating BPM metric

Here, I will create the BPM for each player in each game, since I couldn't find data that pre-computed this. This will then be cumulatively summed over the regular season as a metric to measure performance before and after injury. I'm following the guide on https://www.basketball-reference.com/about/bpm2.html.

```python
# First, we need to calculate position and offensive role for each player
# These would typically require regressions, but for now I'll create placeholder functions
# You'll need to implement the actual position/role calculations based on the appendix regressions


# TODO: Make this correct
def calculate_position_value(player_stats):
    """
    Calculate position value between 1 (PG) and 5 (C)
    This is a placeholder - you'll need the actual regression from the appendix
    """
    # Placeholder logic - replace with actual regression
    return player_stats["position_value"]


def calculate_offensive_role(player_stats):
    """
    Calculate offensive role between 1 (Creator) and 5 (Receiver)
    This is a placeholder - you'll need the actual regression from the appendix
    """
    # Placeholder logic - replace with actual regression
    # return 3.0  # Default to neutral role
    return player_stats["offensive_role"]  # For now, use position as role


def calculate_raw_bpm(row):
    """
    Calculate Box Plus/Minus using the regression coefficients
    """
    # Get position and offensive role values
    pos = calculate_position_value(row)
    off_role = calculate_offensive_role(row)

    # Linear interpolation for position-dependent coefficients
    def interpolate_pos(pos1_val, pos5_val, position):
        return pos1_val + (pos5_val - pos1_val) * (position - 1) / 4

    def interpolate_role(role1_val, role5_val, role):
        return role1_val + (role5_val - role1_val) * (role - 1) / 4

    # Position-dependent coefficients
    pts_coeff = 0.860
    tpm_coeff = 0.389
    to_coeff = -0.964
    fouls_coeff = -0.367

    ast_coeff = interpolate_pos(0.580, 1.034, pos)
    orb_coeff = interpolate_pos(0.613, 0.181, pos)
    drb_coeff = interpolate_pos(0.116, 0.181, pos)
    stl_coeff = interpolate_pos(1.369, 1.008, pos)
    blk_coeff = interpolate_pos(1.327, 0.703, pos)

    # Offensive role-dependent coefficients
    fga_coeff = interpolate_role(-0.560, -0.780, off_role)
    fta_coeff = interpolate_role(-0.246, -0.343, off_role)

    # Calculate BPM
    bpm = (
        pts_coeff * row["points_per_100_possessions"]
        + tpm_coeff * row["3pm_per_100_possessions"]
        + ast_coeff * row["assists_per_100_possessions"]
        + to_coeff * row["turnovers_per_100_possessions"]
        + orb_coeff * row["orb_per_100_possessions"]
        + drb_coeff * row["drb_per_100_possessions"]
        + stl_coeff * row["steals_per_100_possessions"]
        + blk_coeff * row["blocks_per_100_possessions"]
        + fouls_coeff * row["fouls_per_100_possessions"]
        + fga_coeff * row["fga_per_100_possessions"]
        + fta_coeff * row["fta_per_100_possessions"]
    )

    # Add position adjustment constant
    if pos <= 3:
        pos_adjustment = (
            -0.818 * (3 - pos) / 2
        )  # Linear from -0.818 at pos 1 to 0 at pos 3
    else:
        pos_adjustment = 0

    # Add offensive role adjustment constant
    if off_role <= 3:
        role_adjustment = -2.774 + 2.774 * (off_role - 1) / 2  # Linear from -2.774 to 0
    else:
        role_adjustment = 2.774 * (off_role - 3) / 2  # Linear from 0 to 2.774

    return bpm + pos_adjustment + role_adjustment


# Calculate BPM for each game
df["raw_bpm"] = df.apply(calculate_raw_bpm, axis=1)

df[["athlete_display_name", "game_date", "minutes", "raw_bpm"]].sort_values(
    by="raw_bpm", ascending=False
).head(10)
```

```python
# Team Adjustment Calculation - FIXED VERSION
def calculate_team_adjustment(df, season_year):
    """
    Calculate team adjustment for BPM values
    Team adjustment ensures that weighted BPM values sum to team efficiency
    """
    # Filter for regular season games only
    season_df = df.query(f"season == {season_year} and season_type == 2").copy()

    # Calculate team NET efficiency (this should be relative to league average, around 0)
    # We need to calculate both offensive and defensive efficiency
    team_stats = (
        season_df.groupby(["team_id"])
        .agg({"points_team": "mean", "possessions_team": "mean"})
        .reset_index()
    )

    # Calculate league average points per 100 possessions for this season
    league_avg_efficiency = (
        season_df["points_team"].sum() / season_df["possessions_team"].sum() * 100
    )

    # Team efficiency should be relative to league average (typically around 0)
    team_stats["team_efficiency"] = (
        team_stats["points_team"] / team_stats["possessions_team"]
    ) * 100 - league_avg_efficiency

    # Estimate average lead for lead/deficit adjustment
    team_stats["avg_lead"] = 0  # Placeholder - assume neutral for now

    # Apply lead adjustment (teams play 0.35 pts/100 poss worse per point of lead)
    team_stats["lead_adjustment"] = -0.35 / 2 * team_stats["avg_lead"]
    team_stats["adjusted_team_efficiency"] = (
        team_stats["team_efficiency"] + team_stats["lead_adjustment"]
    )

    # Calculate weighted raw BPM sum for each team
    season_df = season_df.merge(
        team_stats[["team_id", "adjusted_team_efficiency"]], on="team_id"
    )

    # Calculate possession percentage for each player on each team
    team_total_possessions = (
        season_df.groupby(["team_id"])["possessions"].sum().reset_index()
    )
    team_total_possessions.columns = ["team_id", "team_total_possessions"]
    season_df = season_df.merge(team_total_possessions, on="team_id")
    season_df["possession_pct"] = (
        season_df["possessions"] / season_df["team_total_possessions"]
    )

    # Calculate weighted raw BPM sum by team
    season_df["weighted_raw_bpm"] = season_df["raw_bpm"] * season_df["possession_pct"]
    team_weighted_bpm = (
        season_df.groupby("team_id")["weighted_raw_bpm"].sum().reset_index()
    )
    team_weighted_bpm.columns = ["team_id", "team_raw_bpm_sum"]

    # Merge with team efficiency
    team_adjustments = team_stats.merge(team_weighted_bpm, on="team_id")

    # Calculate team adjustment constant (should be around -8)
    team_adjustments["team_adjustment"] = (
        team_adjustments["adjusted_team_efficiency"]
        - team_adjustments["team_raw_bpm_sum"]
    )

    return team_adjustments[["team_id", "team_adjustment", "adjusted_team_efficiency"]]


# Recalculate with fixed function
all_team_adjustments = []
for season in df["season"].unique():
    if not pd.isna(season):
        season_adjustments = calculate_team_adjustment(df, int(season))
        season_adjustments["season"] = int(season)
        all_team_adjustments.append(season_adjustments)

team_adjustments_df = pd.concat(all_team_adjustments, ignore_index=True)

# Merge team adjustments back to main dataframe
df = df.merge(team_adjustments_df, on=["team_id", "season"], how="left")
df["adjusted_bpm"] = df["raw_bpm"] + df["team_adjustment"]
df["adjusted_bpm"] = df["adjusted_bpm"].fillna(0)
df
```
